## 1. String Concatenation

```python
str1 = "Hello "
name = input("Enter your Name:")

str2 = str1 + name
print(str2)

str3 = ", welcome to Python programming"
str2 = str2 + str3

print(str2)
```

```python
print("First character:", str2[0])
print("Last character:", str2[-1])
print("First 5 characters:", str2[:5])
print("Last 11 characters:", str2[-11:])
print("Reverse:", str2[::-1])
print("Python:", str2[23:29])
```

```python
strM = "Python beginner tutorial"

print("Uppercase:", strM.upper())
print("Lowercase:", strM.lower())
print("Capitalize:", strM.capitalize())
print("Count of 't':", strM.count('t'))
print("Replace:", strM.replace("Python", "Machine Learning"))
```

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

t_combine = t1 + t2

print("Combined tuple:", t_combine)
print("Repeated 3 times:", t_combine * 3)
print("3rd element:", t_combine[2])
print("First three elements:", t_combine[:3])
print("Last three elements:", t_combine[-3:])
```
